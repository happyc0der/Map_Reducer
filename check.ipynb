```python
# FInd MSE between data points and given set of points
def mse(data, points):
    sum = 0
    for j in range(len(points)):
        for i in range(len(data)):
            sum += (data[i][0] - points[j][0])**2 + (data[i][1] - points[j][1])**2
    return sum
a=[[74.79769897460938, 2.488985300064087, 23.759854157765705], [42.12135553359985, 15.868564128875732, 67.04212760925293]]
b=[[23.86036925, 80.42453215, 20.3360721], [74.24312356, 42.03826089, 35.09856164]]
a=np.array(a).T
b=np.array(b).T
print(a.shape)
print(mse(data,a),mse(data,b))
```
