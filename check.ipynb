```python
# Plot the input points together with the final centroids produced by a run.
import matplotlib.pyplot as plt
import numpy as np

points = np.loadtxt('Data/Input/points.txt', delimiter=',')
centroids = np.loadtxt('Data/centroids.txt', delimiter=',')

plt.scatter(points[:, 0], points[:, 1], label='points')
plt.scatter(centroids[:, 0], centroids[:, 1], color='red', label='centroids')

plt.title('K-Means result')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
plt.legend()
plt.show()
```
